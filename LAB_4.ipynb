```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import split, upper, col

# Create Spark session
spark = SparkSession.builder \
    .appName("TextFileOperations") \
    .getOrCreate()

# a & b. Read text file and create one record per line
# Each line is stored in a String column named "value"
df = spark.read.text("input.txt")

print("Original DataFrame:")
df.show(truncate=False)

# c(i). Selecting a specific column by name
df.select("value").show(truncate=False)

# c(ii). Selecting columns with expressions
# Convert each line to uppercase
df.select(upper(col("value")).alias("uppercase_value")).show(truncate=False)

# Create an array column containing words from each line
df_words = df.withColumn("words", split(col("value"), " "))

df_words.show(truncate=False)

# c(iii). Select the first word from the array column
df_words.select(
    col("words")[0].alias("first_word")
).show(truncate=False)

# Stop Spark session
spark.stop()
```
