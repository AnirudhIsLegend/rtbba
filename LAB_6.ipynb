```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, split, size

# ---------------------------------------------------------
# Create Spark Session
# ---------------------------------------------------------
spark = SparkSession.builder \
    .appName("PySparkFilteringOperations") \
    .getOrCreate()


# ---------------------------------------------------------
# a & b. Ingest text file into DataFrame
# Each line becomes one record.
# The column "value" is of StringType.
# ---------------------------------------------------------
df = spark.read.text("input.txt")

print("Original DataFrame:")
df.show(truncate=False)


# ---------------------------------------------------------
# c(i). Filter rows containing a specific word: "Big"
# ---------------------------------------------------------
print("Rows containing 'Big':")

df.filter(
    col("value").contains("Big")
).show(truncate=False)


# ---------------------------------------------------------
# c(ii). Filtering numeric values
#
# First, split each line into words and count the words.
# ---------------------------------------------------------
word_count_df = df.withColumn(
    "word_count",
    size(split(col("value"), "\\s+"))
)

print("Rows with word count greater than 5:")

word_count_df.filter(
    col("word_count") > 5
).show(truncate=False)


# Words with count between 3 and 10
# between() includes both 3 and 10.
print("Rows with word count between 3 and 10:")

word_count_df.filter(
    col("word_count").between(3, 10)
).show(truncate=False)


# ---------------------------------------------------------
# c(iii). Filtering rows with multiple conditions
# Text must contain BOTH "Big" and "Data"
# ---------------------------------------------------------
print("Rows containing both 'Big' and 'Data':")

df.filter(
    col("value").contains("Big") &
    col("value").contains("Data")
).show(truncate=False)


# ---------------------------------------------------------
# Alternative syntax using where()
# ---------------------------------------------------------
print("Using where() to find rows containing 'Big':")

df.where(
    col("value").contains("Big")
).show(truncate=False)


# Stop Spark Session
spark.stop()
```
