```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, upper, split, explode

# ---------------------------------------------------------
# Create Spark Session
# ---------------------------------------------------------
spark = SparkSession.builder \
    .appName("TextFileDataFrameOperations") \
    .getOrCreate()


# ---------------------------------------------------------
# a & b. Ingest text file into DataFrame
# Each line becomes one record.
# The column "value" is of StringType.
# ---------------------------------------------------------
df = spark.read.text("input.txt")

print("Original DataFrame:")
df.show(truncate=False)


# ---------------------------------------------------------
# c(i). Select column using col() function
# ---------------------------------------------------------
print("Selecting value column using col():")
df.select(col("value")).show(truncate=False)


# ---------------------------------------------------------
# c(ii). Select with column transformation
# Convert the text to uppercase
# ---------------------------------------------------------
print("Text converted to uppercase:")
df.select(
    upper(col("value")).alias("uppercase_value")
).show(truncate=False)


# ---------------------------------------------------------
# c(iii). Explode words
# split() creates an array of words.
# explode() creates a new row for every word.
# ---------------------------------------------------------
words_df = df.select(
    explode(split(col("value"), "\\s+")).alias("word")
)

print("Exploded words:")
words_df.show(truncate=False)


# ---------------------------------------------------------
# d. Order the results and display top 10
# orderBy() sorts the words alphabetically.
# ---------------------------------------------------------
print("Top 10 words in alphabetical order:")

words_df.orderBy(
    col("word").asc()
).show(10, truncate=False)


# Stop Spark session
spark.stop()
```
